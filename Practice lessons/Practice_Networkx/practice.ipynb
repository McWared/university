```python
import networkx as nx
import matplotlib as plt
```

```python
g = nx.Graph()

coordinates = {
    0: (0,0),
    1: (1,0),
    2: (2,0),
    3: (0,1),
    4: (1,1),
    5: (2,1),
    6: (0,2),
    7: (1,2),
    8: (2,2)
}

g.add_edges_from([(0,3), (1,2), (2,5), (5,8), (8,7), (6,7), (6,3), (0,1), (3,4), (4,5), (7,4), (4,1)])
nx.draw(g, node_color = 'aqua', edgecolors = 'blue', with_labels = True, pos = coordinates)
```

```python
nx.write_adjlist(g, 'text.txt')
print(*nx.dfs_edges(g))
print(*nx.bfs_edges(g, source=0))
```

```python
G = nx.path_graph(5)
path = nx.single_source_shortest_path(G, 0)
nx.draw(G, with_labels=True)
print(path)
```
